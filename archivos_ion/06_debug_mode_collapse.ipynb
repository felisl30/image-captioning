```python
# 1. Setup
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

OUTPUT_DIR = PROJECT_ROOT / "outputs" / "mode_collapse_debug"
```

```python
# 2. Comando D1 que se correrá cuando estén los checkpoints
cmd = """
python scripts/run_d1_checkpoint_collapse.py \
  --checkpoint-root models/blip_finetuned_5k \
  --indices data/selected_indices.json \
  --max-images 30 \
  --device cpu
"""
print(cmd)
```

```python
# 3. Leer resultados
summary_path = OUTPUT_DIR / "checkpoint_collapse_summary.csv"
captions_path = OUTPUT_DIR / "checkpoint_captions.csv"
examples_path = OUTPUT_DIR / "checkpoint_examples.csv"

summary_df = pd.read_csv(summary_path)
captions_df = pd.read_csv(captions_path)
examples_df = pd.read_csv(examples_path)

summary_df
```

```python
# 4. Plot simple de collapse
ax = summary_df.plot(
    x="checkpoint",
    y=["unique_ratio", "top_pct"],
    kind="bar",
    figsize=(8, 4),
)

ax.set_title("D1 — Mode collapse por checkpoint")
ax.set_ylabel("Proporción")
ax.set_xlabel("Checkpoint")
plt.xticks(rotation=0)
plt.tight_layout()

fig_path = OUTPUT_DIR / "checkpoint_collapse_plot.png"
plt.savefig(fig_path, dpi=200)
plt.show()

print("Guardado en:", fig_path)
```

```python
# 5. Ejemplos cualitativos
for checkpoint, group in examples_df.groupby("checkpoint"):
    print("=" * 100)
    print("CHECKPOINT:", checkpoint)
    display(group[["kind", "idx", "reference", "caption"]])
```
